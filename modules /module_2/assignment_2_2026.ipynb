# Copy new files from a server
Let's say you are working at a county board where field personnel are collecting UAV imagery almost every day. You are tasked with processing these images, so you want to copy them to your own workstation daily. This can, of course, be done by navigating the file tree in Windows and manually copying the new files. However, you are a bit lazy and decide to write a Python script to do it for you.

```python
import os
import shutil
import datetime
import time
import glob

def copy_files(source_dir, destination_dir):
  # Create a new directory but only if it does not already exist.
  os.makedirs(destination_dir, exist_ok=True)
  # Create a list of files in the destination directory
  list_of_old_files = []
  for file_name in os.listdir(destination_dir):
    list_of_old_files.append(file_name)
  # Get the list of files in the source directory
  list_of_new_files = []
  for file_name in os.listdir(source_dir):
    if file_name not in list_of_old_files:
      # merge paths with the file_names
      source_file_path = os.path.join(source_dir, file_name)
      destination_file_path = os.path.join(destination_dir, file_name)
      # Finally copy files
      shutil.copy(source_file_path, destination_file_path)
      print(f"Copied {file_name} to {destination_dir}")

# Run the backup function
source_directory = '/content/sample_data' # Paths from google colab
destination_directory = '/content/copied_files' # Paths from google colab
copy_files(source_directory, destination_directory)
```

# Webscraping
The web is full of usefull information and we can use python to parse it and find patterns. BeautifulSoup is one package that can parse text on websites. this can be used to search for specific words or phrases. In the example bellow we search for how many times "research" appears on the SLU webpage.

```python
from bs4 import BeautifulSoup
import requests

# Fetch the HTML content from a URL
url = 'https://www.slu.se/en/'
response = requests.get(url)
html_content = response.text

# Parse the HTML content with BeautifulSoup
soup = BeautifulSoup(html_content, 'html.parser')

# Extract information by navigating the HTML structure
title = soup.title.text
paragraphs = soup.find_all('p')

# Search for a specific word or phrase
search_term = 'research'
occurrences = 0

# Print the extracted information and count occurrences
print(f'Title: {title}')
for i, p in enumerate(paragraphs, 1):
    paragraph_text = p.text
    print(f'Paragraph {i}: {paragraph_text}')

    # Check if the search term is present in the paragraph
    if search_term.lower() in paragraph_text.lower():
        occurrences += 1

# Print the count of occurrences
print(f'The word "{search_term}" appears {occurrences} times on the webpage.')
```

### Scrape images
We can also download images from websites using python. Here is an example using beautifulsoup4

```python
!pip install beautifulsoup4 requests
```

```python
import requests
from bs4 import BeautifulSoup
import os
import urllib.parse

def download_images(url, folder="images"):
    os.makedirs(folder, exist_ok=True)
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    # Find all image tags
    img_tags = soup.find_all('img')
    img_urls = [img['src'] for img in img_tags if 'src' in img.attrs]

    for img_url in img_urls:
        # Handle relative URLs
        img_url = urllib.parse.urljoin(url, img_url)
        try:
            img_data = requests.get(img_url).content
            img_name = os.path.join(folder, os.path.basename(img_url))
            with open(img_name, 'wb') as f:
                f.write(img_data)
            print(f"Downloaded {img_name}")
        except Exception as e:
            print(f"Failed to download {img_url}: {e}")

# Example usage
download_images("https://www.slu.se/")
```

```python
import matplotlib.pyplot as plt
import os
from PIL import Image
import io
import cairosvg
import shutil

# Folder with downloaded images
folder = "/content/images"

# Step 1: Clean filenames (remove query parameters)
for f in os.listdir(folder):
    if "?" in f:
        clean_name = f.split("?")[0]
        os.rename(os.path.join(folder, f), os.path.join(folder, clean_name))

# Step 2: List first N images to display
images = os.listdir(folder)[:6]  # Adjust number as needed

plt.figure(figsize=(15, 5))
for i, img_file in enumerate(images):
    img_path = os.path.join(folder, img_file)
    ext = img_file.lower().split('.')[-1]

    # Handle SVG images
    if ext == "svg":
        png_data = cairosvg.svg2png(url=img_path)
        img = Image.open(io.BytesIO(png_data))
    else:
        img = Image.open(img_path)

    plt.subplot(1, len(images), i + 1)
    plt.imshow(img)
    plt.axis("off")
    plt.title(img_file)

plt.tight_layout()
plt.show()
```
